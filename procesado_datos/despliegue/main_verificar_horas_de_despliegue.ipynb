```python
# Crear la instancia del manager de configuraciones
config = ProcesadoConfig.from_sources(
    config_modulo_procesado,
    config_despliegue,
)    
```

```python
ruta_a_carpeta_de_datos_crudos = config.carpeta_de_datos_crudos
for serial in config.seriales_de_sondas:
    ruta_al_archivo = os.path.join(ruta_a_carpeta_de_datos_crudos, f"datos_Localizacion_{serial}_TOTAL.csv")
    df = pd.read_csv(ruta_al_archivo)
    if df.empty:
        print(f"No hay datos para la sonda {serial}")
        continue
    latitud = df["latitud"].iloc[0]
    longitud = df["logitud"].iloc[0]
    fecha = df["fecha"].iloc[0]
    fecha = pd.to_datetime(fecha, format = "%Y-%m-%dT%H:%M:%S.%fz")
    fecha_primera_medicion_str = fecha.strftime("%Y-%m-%d %H:%M")
    fecha_despliegue_str = fecha.strftime("%d/%m/%Y %H:%M")
    fecha_maniobra = fecha-pd.Timedelta(minutes=2)
    fecha_maniobra_str = fecha_maniobra.strftime("%d/%m/%Y %H:%M")
    print(f"*** BOYA ***")
    print(serial)
    print(f"Latitud: {latitud}")
    print(f"Longitud: {longitud}")
    print(f"Fecha y hora de la primera medicion [UTC]: {fecha_primera_medicion_str}")
    print(f"Fecha_despliegue [UTC]: {fecha_despliegue_str}")
    print(f"Fecha maniobra [UTC]: {fecha_maniobra_str}")
    print("")
```
