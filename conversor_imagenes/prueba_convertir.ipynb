```python
import os
from PIL import Image
from pillow_heif import register_heif_opener

register_heif_opener()

def convertir(directorio_raiz):

    contador = 0

    for ruta_actual, _, archivos in os.walk(directorio_raiz):

        for archivo in archivos:

            if archivo.lower().endswith(".heic"):

                ruta_heic = os.path.join(ruta_actual, archivo)

                nombre_base = os.path.splitext(archivo)[0]
                ruta_png = os.path.join(ruta_actual, nombre_base + ".png")

                try:
                    print(f"Abriendo: {ruta_heic}")

                    imagen = Image.open(ruta_heic)

                    if imagen.mode != "RGB":
                        imagen = imagen.convert("RGB")

                    imagen.save(ruta_png)

                    contador += 1

                    print(f"OK -> {ruta_png}")

                except Exception as e:
                    print(f"ERROR -> {ruta_heic}")
                    print(e)

    print(f"\nTotal convertidos: {contador}")

convertir(r"C:\Users\Atmosfera\Documents\Doris_Isla_Aguada_26mayo2026")
```
